```python
from temp_email import TempEmail
from register import fill_registration_form

# Init desired number of API keys
NUM_API_KEYS = 1000
api_keys = []
total_attempts = 0
num_recent_keys = 0
success_rates = []

# change to cycle between domains
# prolly can't keep using same domain after while
# Generate API keys using temporary emails
while len(api_keys) < NUM_API_KEYS:    
    total_attempts += 1
    if total_attempts % 10 == 0:
        success_rates.append(num_recent_keys / 10)

    email = TempEmail()

    if not email.token:
        print("Skipping this email due to missing authentication token.")
        continue

    fill_registration_form(email.email_address)
    api_key = email.fetch_api_key()
    if api_key:
        api_keys.append(api_key)
        num_recent_keys += 1
    else:
        print("⚠️ No API key found, skipping.")

# Write API keys to a file for storage
with open(f"data/keys_{NUM_API_KEYS}.txt", "w") as file:  
    file.write("\n".join(api_keys)) 

# Record success rate
print(NUM_API_KEYS // total_attempts)
```
