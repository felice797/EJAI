```python
from comments_downloader import CommentsDownloader

def load_data_for_document(key, document_id):
    downloader = CommentsDownloader(api_key=key)
   
    print(f"\n********************************\nSTARTING {document_id}\n********************************")
    downloader.gather_comments_by_document(document_id, csv_filename=f"documents/{document_id}.csv")

    print("\nDONE")

```

```python
import pandas as pd

df = pd.read_csv("output_with_documents.csv")  # Replace with your actual file name

# Get list from specific column
documents = df["document_id"].dropna().tolist()  # replace "column_name" with your column

print(documents)
print(len(documents))
```

```python
from concurrent.futures import ThreadPoolExecutor
from tqdm import tqdm

def load_wrapper(args):
    key, document_id = args
    load_data_for_document(key, document_id)

assert len(documents) <= len(keys), "Need at least one API key per document."

with ThreadPoolExecutor(max_workers=10) as executor:
    list(tqdm(
        executor.map(load_wrapper, zip(keys, documents)),
        total=len(documents)
    ))
```
