```python
from comments_downloader import CommentsDownloader

def load_data_for_document(key, document):
    downloader = CommentsDownloader(api_key=key)

    my_documents = document

    for document_id in my_documents:
        print(f"\n********************************\nSTARTING {document_id}\n********************************")
        downloader.gather_comments_by_document(document_id, csv_filename=f"mydocuments/{document_id}.csv")

    print("\nDONE")

```
